```python
import sys
sys.path.append("..")

import pandas as pd
from sklearn.model_selection import train_test_split
from src.preprocessing import get_preprocessor

df = pd.read_csv("../data/telco_clean.csv")
df.shape
```

```python
X = df.drop(columns=["Churn"])
y = df["Churn"]
print(X.shape, y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(X_train.shape, X_test.shape)
print("Train churn rate:", round(y_train.mean(), 3))
print("Test churn rate:", round(y_test.mean(), 3))
```

```python
preprocessor = get_preprocessor(X_train)

X_train_prep = preprocessor.fit_transform(X_train)
X_test_prep = preprocessor.transform(X_test)

print(X_train_prep.shape, X_test_prep.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()
print(len(feature_names))
print(feature_names)

prep_df = pd.DataFrame(X_train_prep, columns=feature_names)
prep_df[["num__tenure", "num__MonthlyCharges", "num__TotalCharges"]].describe().loc[["mean", "std"]].round(2)
```

```python
X_train.assign(Churn=y_train).to_csv("../data/train.csv", index=False)
X_test.assign(Churn=y_test).to_csv("../data/test.csv", index=False)
```
